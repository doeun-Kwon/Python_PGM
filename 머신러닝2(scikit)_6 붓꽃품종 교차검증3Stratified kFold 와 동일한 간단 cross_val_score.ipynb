```python
import sklearn
import pandas as pd
import numpy as np

#from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score, cross_validate

########################################################################
#
# cross_val_score :
# 기존 StratifiedKFold 를 이용한 데이터분포반영/훈련 및 검증데이터 분리/예측 을 
# 하나의 함수로 구현한 것 
#
########################################################################
iris = sklearn.datasets.load_iris()
iris_data = iris.data
iris_label = iris.target

iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)
iris_df['label'] = iris.target

print('\n++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++++')
print('\nCross_val_score')

print('\nDT_clf')
DT_clf = DecisionTreeClassifier(random_state=156)

scores = cross_val_score(DT_clf, iris_data, iris_label, scoring='accuracy', cv=3)

print('\n교차 검증정확도 : ',np.round(scores,4))
print('\n평균검증정확도 : ', np.round(np.mean(scores),4))
    
```
