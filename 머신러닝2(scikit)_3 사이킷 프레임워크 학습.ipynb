```python
import sklearn
import pandas as pd
import numpy as np

from sklearn.datasets import load_iris
#from sklearn.datasets import load_boston #<-- 사용금지
from sklearn.datasets import load_breast_cancer
from sklearn.datasets import load_diabetes
from sklearn.datasets import load_digits
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

iris = load_iris()
brc = load_breast_cancer()
dia = load_diabetes()
dig = load_digits()

iris_data = iris.data
iris_label = iris.target

#학습/테스트 데이터 분할
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label, test_size=0.2, random_state=11)

#분류모델 객체 생성
dt_clf = DecisionTreeClassifier(random_state=11)

#학습수행
dt_clf.fit(X_train, y_train)

#예측수행
pred = dt_clf.predict(X_test)

#학습결과 출력
print('\n예측 정확도 : {0: .4f}' .format(accuracy_score(y_test, pred)))

print('\n')
print(iris.data)
print(iris.data.shape)
print(iris.feature_names)
print(iris.target)
print(iris.target.shape)
print(iris.target_names)
print('----------------------')


'''
print(brc.data)
print(brc.feature_names)
print(brc.target)
print(brc.target_names)
print('----------------------')
print(dia.data)
print(dia.feature_names)
print(dia.target)
#print(dia.target_names) <-- 에러 발생. 원인 불명
print('----------------------')
print(dig.data)
print(dig.feature_names)
print(dig.target)
print(dig.target_names)
'''
```
