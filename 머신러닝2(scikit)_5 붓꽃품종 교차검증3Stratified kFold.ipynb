```python
import sklearn
import pandas as pd
import numpy as np

#from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score

########################################################################
#
# 훈련과 검증 셋트를 n회 반복적으로 구성하여 학습하는 방식 (StratifiedKFold)
# 데이터의 결과값(label) 비율을 훈련 및 검증데이터 모두 폴드 비율로 구성하는 것
#
########################################################################
iris = sklearn.datasets.load_iris()
iris_data = iris.data
iris_label = iris.target

iris_df = pd.DataFrame(data=iris_data, columns=iris.feature_names)
iris_df['label'] = iris.target

print('\nDT_clf')
DT_clf = DecisionTreeClassifier(random_state=156)

skfold = StratifiedKFold(n_splits=3)
cv_accuracy = []

print('\n데이터 크기 : ', iris_data.shape[0])

n_iter = 0
for train_index, test_index  in skfold.split(iris_df, iris_df['label']):
    X_train, X_test = iris_data[train_index], iris_data[test_index]
    y_train, y_test = iris_label[train_index], iris_label[test_index]

    iris_train = iris_df['label'].iloc[train_index]
    iris_test = iris_df['label'].iloc[test_index]
    
    DT_clf.fit(X_train, y_train)
    pred = DT_clf.predict(X_test)

    n_iter += 1
    accuracy = np.round(accuracy_score(y_test, pred), 4)
    train_size = X_train.shape[0]
    test_size = X_test.shape[0]
    print('\n========================================================================')
    print('\n#{0} 교차검증' .format(n_iter))
    print('\n▶ 훈련셋트 인덱스 #', train_index)
    print('\n▶ 훈련셋트 데이타분포 #', iris_train.value_counts())
    print('\n▶ 검증셋트 인덱스 #', test_index)
    print('\n▶ 검증셋트 데이터분포 #', iris_test.value_counts())
    print('\n==>  교차검증정확도 : {1}, 학습데이터 크기 : {2},  검증데이터 크기 : {3}'
          .format(n_iter, accuracy, train_size, test_size))
    
    cv_accuracy.append(accuracy)

print('\n교차 검증정확도 : ',np.round(cv_accuracy,4))
print('\n평균검증정확도 : ', np.mean(cv_accuracy))
    
```
