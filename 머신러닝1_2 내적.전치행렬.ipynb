```python
import numpy as np

A = np.array([[1,2,3],
              [4,5,6]])
B = np.array([[7,8],
              [9,10],
              [11,12]])
A_dot_B = np.dot(A, B)
print('A 내적 B (np.dot(A, B)) : \n', A_dot_B)

A_transpose = np.transpose(A)
print('\nA 의 전치행렬은 (np.transpose(A)) : \n', A_transpose)
```
