`````````python
import pandas as pd
import numpy as np

# [데이터 프레임 연습] ###

# DataFrame 만들기
#   -- 1. list로부터 만들기

col_name = ['나이','성별', '주소']
list1 = [[13, 'Male', '경북 포항'],
         [24, 'Female', '부산'],
         [34, 'Male', '서울']]
print('#리스트 :\n', list1)
df_list = pd.DataFrame(list1, columns=col_name)
print('\n#리스트로부터 만든 데이터프레임 :\n', df_list)

#   -- 2. array로부터 만들기
arr1 = np.array(list1)
df_array = pd.DataFrame(arr1, columns=col_name)
print('\n#array로부터 만든 데이터프레임 :\n', df_array)

#   -- 3. dictionary로부터 만들기
dict1 = {'나이':[13,24,34], '성별':['Male','Female','Male'], '주소':['경북 포항','부산','서울']}
df_dict = pd.DataFrame(dict1)
print('\n#Dictionary로부터 만든 데이터프레임 :\n', df_dict)

# csv 파일 읽기 
#   -- 아래 5개의 read_csv 는 모두 같은 read 기능을 한다.

#titanic_df = pd.read_csv(r'..\Python_Data\titanic_train.csv')
titanic_df = pd.read_csv(r'C:\Users\dekwo\Python_Data\titanic_train.csv')
#titanic_df = pd.read_csv(r'../Python_Data/titanic_train.csv')
#titanic_df = pd.read_csv('..//Python_Data//titanic_train.csv')
#titanic_df = pd.read_csv('..\\Python_Data\\titanic_train.csv')
#titanic_df = pd.read_csv("../Python_Data/titanic_train.csv", sep=',')

# pd.read_fwf : Fixed Width 고정 길이 컬럼을 읽을 때 사용

print('\n#titanic_df.head(3) : after Age added')
titanic_df.head(3)    # <---- 이 것은 표시가 안된다. 이유를 모르겠다`````` 이 파일의 맨 마지막 head 만 실행된다  ````````

print('\n#Data frame의 크기 (titanic_df.shape): ', titanic_df.shape)
print('\n#titanic_df.info() : ')
titanic_df.info()
print('\n#titanic_df.describe() :')
titanic_df.describe()
val_cnt = titanic_df['Sex'].value_counts()
print('\n#titanic_df[''Sex''].value_counts() : (value의 종류별 카운트를 정리) : \n', val_cnt)

titanic_df_sex = titanic_df['Sex']
print('\n#titanic_df[''Sex''] 의 리스트 \n', titanic_df_sex)
val_cnt2 = titanic_df_sex.value_counts()
print('\n#titanic_df_sex.value_counts() \n', val_cnt2)

titanic_df['Age_10'] = titanic_df['Age'] + 10

print('\n#titanic_df.head(3) : after Age added')
titanic_df.head(3)

#컬럼에 함수를 적용하여 계산
print('\ntitanic_df[''Age''].apply(lambda x : x**2) ==> 나이 제곱 결과')
titanic_df['Age_**2'] = titanic_df['Age'].apply(lambda x : x**2)
titanic_df.head(3)


#컬럼 삭제
print('\ntitanic_df_drop = titanic_df.drop(''Age_10'', axis=1) ==> titanic_df 의 Age_10 컬럼은 그대로 유지됨.')
titanic_df_drop = titanic_df.drop('Age_10', axis=1)
titanic_df.head(3)
print('\ntitanic_df_drop = titanic_df.drop(''Age_10'', axis=1, inplace=True) ==> titanic_df 의 Age_10 컬럼도 삭제됨.')
titanic_df_drop = titanic_df.drop('Age_10', axis=1, inplace=True)
titanic_df.head(3)
`````````
