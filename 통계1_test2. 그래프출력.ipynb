```python
import array
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import random

ar = array.array('i', [300,200,100])

ay = np.array([100,100,100])
ar1 = ar
ay1 = ay
res = ar1 * ay1
print("ar:"); print(ar)
print("ay:"); print(ay)
print("res:"); print(res)
print("\n")


#########

#random.seed(20)
#np.random.seed(21)
user_id = np.random.randint(100000,200000,10)

price_list = random.sample(range(100,1000), 10)
price_list = [x*1000 for x in price_list]

reduce_price_list = random.sample(range(100,500), 10)
reduce_price_list = [x*10 for x in reduce_price_list]

act_price_list = [x - y for x, y in zip(price_list, reduce_price_list)]

vip_array = np.random.choice(["Y", "N"], size=10, p=[0.3, 0.7])

# Dictionary 선언
df_dict = {
    '고객id': user_id,
    '구매날짜': pd.Timestamp('20250930'),
    '구매가격': price_list,
    '할인가격': reduce_price_list,
    '실제가격': act_price_list,
    'VIP' : vip_array }

# 딕셔너리를 DataFrame 으로 전환
df = pd.DataFrame(df_dict)


print("구매가격")
print(price_list)
df['구매가격'].plot()
plt.show()

print("할인가격")
print(reduce_price_list)
df['할인가격'].plot()
plt.show()

print("실제가격")
print(act_price_list)
df['실제가격'].plot()
plt.show()

df['실제가격2'] = df['구매가격'] - df['할인가격']

print("실제가격2")
print(df['실제가격2'])
df['실제가격2'].plot()
plt.show()

x = np.arange(0, 5, 0.1)
y = np.sin(x)
plt.plot(x, y)
plt.show()

x = np.arange(0, 10, 0.1)
y = np.sin(x)
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title(f'File store & read time \n Dataframe Size')
plt.show()

```
