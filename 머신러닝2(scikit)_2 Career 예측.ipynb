```python
import sklearn
import pandas as pd
import numpy as np

#from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

print(sklearn.__version__)
car_df = pd.read_csv("../Python_Data/choice_Career.csv")

data_df = car_df.iloc[:,:-1]
result_df = car_df.iloc[:,-1]
print(data_df)
print(result_df)

print('\nX_train , X_test , y_train , y_test  (학습 데이터세트, 테스트 데이터세트, 학습 라벨, 테스트 라벨)')
X_train, X_test, y_train, y_test = train_test_split(data_df, result_df, test_size=0.2, random_state=11)
print('####### 여기까지 데이터 셋팅 완료 #######\n')

print('\n머신러닝 함수 선언 DT_clf = DecisionTreeClassifier(random_state=11)')
DT_clf = DecisionTreeClassifier(random_state=11)

print('\n학습 수행 DT_clf.fit(X_train, y_train)')
DT_clf.fit(X_train, y_train)

print('\n예측 수행 pred = DT_clf.predict(X_test)')
pred = DT_clf.predict(X_test)
print('\n예측 정답 y_test')
print(y_test)

print('\n예측 결과 pred')
print(pred)
print('28개 정답 / 총 30개 =  {0: .4f} %'.format(28/30))
print('\n예측 판정(정확도, accuracy_score(y_test, pred)) ) : {0: .4f} %'.format(accuracy_score(y_test, pred)))
```
